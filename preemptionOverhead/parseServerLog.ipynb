```python
import re

# Arrays to store the extracted numbers
queue_numbers = []
reply_numbers = []

# Regular expressions for matching the required lines
pattern_queue = re.compile(r'putting reply to replyHandlerQueue (\d+) : (\d+)')
pattern_reply = re.compile(r'reply (\d+) : (\d+)')

# Open and read the file line by line
with open('noPreemption_3mode_woLengthControl_log.txt', 'r') as file:
    for line in file:
        match_queue = pattern_queue.search(line)
        if match_queue:
            number_before_colon = int(match_queue.group(1))
            number_after_colon = int(match_queue.group(2))
            if number_before_colon < 100:
                queue_numbers.append(number_after_colon)
            continue

        match_reply = pattern_reply.search(line)
        if match_reply:
            number_before_colon = int(match_reply.group(1))
            number_after_colon = int(match_reply.group(2))
            if number_before_colon < 100:
                reply_numbers.append(number_after_colon)

# Now queue_numbers and reply_numbers contain the extracted numbers
print("Queue Numbers:", queue_numbers)

print(len(queue_numbers))
print("Reply Numbers:", reply_numbers[130:])
print(len(reply_numbers[130:]) )

import numpy as np
reply_latency = np.array(reply_numbers[130:]) - np.array(queue_numbers)
print(reply_latency)
print(np.mean(reply_latency))
```
